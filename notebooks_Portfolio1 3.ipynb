```python
# Daily Budget Tracker

print("DAILY BUDGET TRACKER\n")

name = input("Enter your name: ")
period = input("Track budget for Week or Month? ").lower()

# Number of days
if period == "week":
    days = 7
elif period == "month":
    days = 30
else:
    print("Invalid input. Defaulting to Week.")
    days = 7

allowance = float(input("Enter your total allowance: "))

# Totals
total_food = 0
total_transpo = 0
total_misc = 0

# Daily expenses
for day in range(1, days + 1):
    print(f"\nDay {day}")

    food = float(input("Food expense: "))
    transpo = float(input("Transportation expense: "))
    misc = float(input("Miscellaneous expense: "))

    total_food += food
    total_transpo += transpo
    total_misc += misc

# Computations
total_expenses = total_food + total_transpo + total_misc
balance = allowance - total_expenses
average = total_expenses / days

# Output
print("\n===== BUDGET SUMMARY =====")

print(f"Name: {name}")
print(f"Tracking Period: {period}")

print(f"\nTotal Food Expenses:           {total_food:>8.2f}")
print(f"Total Transportation Expenses: {total_transpo:>8.2f}")
print(f"Total Miscellaneous Expenses:  {total_misc:>8.2f}")

print(f"\nTotal Expenses:                {total_expenses:>8.2f}")
print(f"Remaining Balance:             {balance:>8.2f}")
print(f"Average Daily Spending:        {average:>8.2f}")

# Status
if balance > 0:
    print("\nStatus: You are within your budget.")
elif balance == 0:
    print("\nStatus: Sakto lang ang budget.")
else:
    print("\nStatus: You are over budget.")
```
